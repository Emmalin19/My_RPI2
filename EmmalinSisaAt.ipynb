```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
%matplotlib inline
```

```python
x=np.array([1,2,3,4,5])
y=np.array([5,4,5,3,2])
plt.xlim(0,8)
plt.ylim(0,8)
plt.scatter(x,y)
plt.title("Scatter Plot")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
